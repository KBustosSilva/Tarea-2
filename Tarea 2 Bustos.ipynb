```python
# Problema 1. Clasificador de temperatura

# Aparecen los números del -30 hasta el 50 (el 51 de abajo no cuenta)
for temperatura in range(-30, 51):

# Si la temperatura es menor que 0
  if temperatura < 0:
    print(temperatura, "Muy frío")

# Si la temperatura está entre 0 y 24
  elif temperatura >= 0 and temperatura <= 24:
    print(temperatura, "Templado")

# Si la temperatura es mayor o igual a 25
  else:
    print(temperatura, "Caluroso")
```

```python
# Problema 2. Números especiales del robot

# Aparecen los números del 1 al 100
for numero in range(1, 101):

# Revisamos los números divisibles por 10
  if numero % 10 == 0:
    print(numero, "Bip Bip")

# Si no lo es, revisamos si es divisible por 5
  elif numero % 5 == 0:
    print(numero, "Bip")
```

```python
# Problema 3. Detector de puntaje gamer

# Solicitamos la entrada de los dos puntajes al usuario (en valores enteros)
puntaje1 = int(input("Ingresa primer puntaje: "))
puntaje2 = int(input("Ingresa segundo puntaje: "))

# Sumamos los dos puntajes
suma = puntaje1 + puntaje2

#Evaluamos las sumas según los criterios
if suma < 50:
  print("Resultado: Nivel principiante")

elif suma >= 50 and suma < 80:
  print("Resultado: Nivel intermedio")

else:
  print("Resultado: Nivel pro")
```

```python
# Problema 4. Calculador de perfil musical

# Variables de entrada
edad = int(input("Ingresa tu edad: "))
le_gusta_musica = input("¿Te gusta la música? (si/no): ") == "si"
le_gusta_bailar = input("¿Te gusta bailar? (si/no): ") == "si"

# Si es mayor de edad, le gusta la música y le gusta bailar
if edad >= 18 and le_gusta_musica and le_gusta_bailar:
  print("Eres una estrella de la fiesta")

# Si le gusta la música pero no le gusta bailar
elif le_gusta_musica and not le_gusta_bailar:
  print("Te gusta la música, pero no tanto bailar")

# Si no cumple ninguna de las condiciones anteriores
else:
  print("Tienes tu propio estilo")
```
